```python
customers_df = (spark.read
                .option("header", "true")
                .option("inferSchema", "true")
                .csv("/Volumes/workspace/default/retail_data/customer.csv"))
products_df = (spark.read
                .option("header", "true")
                .option("inferSchema", "true")
                .csv("/Volumes/workspace/default/retail_data/products.csv"))
orders_df = (spark.read
                .option("header", "true")
                .option("inferSchema", "true")
                .csv("/Volumes/workspace/default/retail_data/orders.csv"))
                
display(products_df)
display(orders_df)
display(customers_df)
```

```python
customers_df.write.mode("overwrite").format("delta").saveAsTable("bronze_customers")
products_df.write.mode("overwrite").format("delta").saveAsTable("bronze_products")
orders_df.write.mode("overwrite").format("delta").saveAsTable("bronze_orders")

spark.sql("SHOW TABLES").show()
```
