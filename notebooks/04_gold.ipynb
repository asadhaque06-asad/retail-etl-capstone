```python
silver_df = spark.table("silver_enriched_orders")

silver_df.printSchema()
silver_df.show()
```

```python
from pyspark.sql.functions import sum, count
gold_sales = (
    silver_df.groupBy("order_date")
    .agg(
        sum("revenue").alias("total_revenue"),
        count("order_id").alias("total_orders"),
    )
)
```

```python
gold_sales.orderBy("order_date").show()
```

```python
gold_sales.write.mode("overwrite").format("delta").saveAsTable("gold_sales_summary")
```

```python
spark.table("gold_sales_summary").show(5)
```

```python
from pyspark.sql.functions import sum, col

gold_customers = (
    silver_df.groupBy("customer_id", "customer_name")
    .agg(
        sum("revenue").alias("total_spend"))
    .orderBy(col("total_spend").desc())
)
gold_customers.show(10)
```

```python
gold_customers.write \
    .mode("overwrite") \
    .format("delta") \
    .saveAsTable("gold_top_customers")
```

```python
spark.catalog.listTables()
```

```python
gold_region =(
    silver_df.groupBy("region")
    .agg(
        sum("revenue").alias("total_revenue")
        )
    .orderBy("total_revenue", ascending=False)
)
gold_region.show()
```

```python
gold_region.write \
    .mode("overwrite") \
    .format("delta") \
    .saveAsTable("gold_region_sales")
```
