```python
customers_df = (spark.read
                .option("header", "true")
                .option("inferSchema", "true")
                .csv("/Volumes/workspace/default/retail_data/customer.csv"))
products_df = (spark.read
                .option("header", "true")
                .option("inferSchema", "true")
                .csv("/Volumes/workspace/default/retail_data/products.csv"))
orders_df = (spark.read
                .option("header", "true")
                .option("inferSchema", "true")
                .csv("/Volumes/workspace/default/retail_data/orders.csv"))
```

```python
customers_df.printSchema()
products_df.printSchema()
orders_df.printSchema()
```

```python
customers_df.show(5, truncate=False)
products_df.show(5, truncate=False)
orders_df.show(5, truncate=False)
display(customers_df)
display(products_df)
display(orders_df)
```

```python
from pyspark.sql.functions import col, sum, when

customers_df.select([sum(when(col(c).isNull(), 1).otherwise(0)).alias(c)
                     for c in customers_df.columns]).show()
products_df.select([sum(when(col(c).isNull(), 1).otherwise(0)).alias(c)
                    for c in products_df.columns]).show()
orders_df.select([sum(when(col(c).isNull(), 1).otherwise(0)).alias(c)
                  for c in orders_df.columns]).show()
```

```python
print("customer duplicate:", customers_df.count() - customers_df.dropDuplicates(['customer_id']).count())
print("product duplicate:", products_df.count() - products_df.dropDuplicates(['product_id']).count())
print("order duplicate:", orders_df.count() - orders_df.dropDuplicates(['order_id']).count())
```
