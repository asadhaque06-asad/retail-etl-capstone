```python
from pyspark.sql import Row

updates = spark.createDataFrame([
    Row(
        order_id=1,
        customer_id=1,
        customer_name='John Lowery',
        region='East',
        product_id=1,
        product_name='Laptop',
        category='Electronics',
        quantity=5,
        unit_price=1200.0,
        revenue=6000.0,
        order_date='2026-07-30',
        status='Completed',
        signup_date='2024-01-01'
    ),
    Row(
        order_id=1000,
        customer_id=2,
        customer_name='Asad',
        region='West',
        product_id=2,
        product_name='Mouse',
        category='Accessories',
        quantity=3,
        unit_price=50.0,
        revenue=150.0,
        order_date='2026-07-30',
        status='Completed',
        signup_date='2024-01-15'
    )
])
```

```python
from delta.tables import DeltaTable

delta_table = DeltaTable.forName(spark, "silver_enriched_orders")
delta_table.alias("t").merge(
    updates.alias("s"),
    "t.order_id = s.order_id") \
        .whenMatchedUpdateAll() \
        .whenNotMatchedInsertAll() \
        .execute()

```
